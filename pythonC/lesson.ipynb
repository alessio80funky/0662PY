```python
#Urlopenを使って、GoogleスプレッドシートのCSVデータを取得する
from urllib.request import urlopen
#CSVデータを扱うための機能
import csv 
#文字列を扱うための機能
import io
```

```python
#urlopenでGoogleスプレッドシートのCSVデータを取得する
url = "https://docs.google.com/spreadsheets/d/1w9xvco8sbc1RCDZvvuKWkzhZ0n1Drlr4dU0oOTktK_Y/edit?usp=sharing"
```

```python
#データを取得する
with urlopen(url) as response:
    #データを文字列として読み込む
    text = response.read().decode("utf-8")
```

```python
#CSVデータを扱うためのDictReaderを使って、文字列からCSVデータを読み込む
reader = csv.DictReader(io.StringIO(text))
```

```python
#データをリストに変換する
data = list(reader)
```
