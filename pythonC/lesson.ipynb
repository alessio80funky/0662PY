```python
#Urlopenを使って、GoogleスプレッドシートのCSVデータを取得する
from urllib.request import urlopen
#CSVデータを扱うための機能
import csv 
#文字列を扱うための機能
import io
```

```python
#Googleスプレッドシートの共有URL
share_url = "https://docs.google.com/spreadsheets/d/1w9xvco8sbc1RCDZvvuKWkzhZ0n1Drlr4dU0oOTktK_Y/edit?usp=sharing"
```

```python
#記述内で「/edit?usp=sharing」を「/export?format=csv」に置き換えることで、CSVデータとして取得できるURLを作成する
url = share_url.split("/edit")[0] + "/export?format=csv"
print(url)
```

```python
#データを取得する
with urlopen(url) as response:
    #データを文字列として読み込む
    text = response.read().decode("utf-8")
```

```python
#CSVデータを扱うためのDictReaderを使って、文字列からCSVデータを読み込む
reader = csv.DictReader(io.StringIO(text))
```

```python
#データをリストに変換する
data = list(reader)
```

```python
total = 0

for row in data:
    #商品名を取り出す
    product = row["商品"]
    #売上を取り出すそして整数に変換する
    sales = int(row["売上"])
    #日付を取り出す
    date = row["日付"]
    #データを表示する
    print(f"{date},{product}の売上は{sales}です。")
    #合計売上を計算する
    total = total + sales

print(f"合計売上は{total}です。")
```
